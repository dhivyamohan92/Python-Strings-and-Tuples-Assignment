```python
# String 1
string_1 = "Hello"

# Get String 2 from the user
string_2 = input("Enter your Name: ")

# Concatenate String 1 and String 2
Final = string_1 + " " + string_2
print(Final)

# String 3
string_3 = "Welcome to Python programming"

# Concatenate String 3 to the existing string
Final = Final + ", " + string_3

print(Final)
```

```python
# Create the concatenated string

string_1 = "Hello"
string_2 = input("Enter your Name: ")

Final = string_1 + " " + string_2

string_3 = "welcome to Python programming"
Final = Final + ", " + string_3

print("Concatenated string:", Final)

# a. Print the first character
print("a. First character:", Final[0])

# b. Print the last character
print("b. Last character:", Final[-1])

# c. Print the first 5 characters
print("c. First 5 characters:", Final[:5])

# d. Print the last 11 characters
print("d. Last 11 characters:", Final[-11:])

# e. Print the string in reverse
print("e. Reverse:", Final[::-1])

# f. Print the word "Python"
print("f. Python:", Final[Final.index("Python"):Final.index("Python") + 6])
```
